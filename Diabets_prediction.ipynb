# Diabetes Prediction Project

## Introduction
Diabetes mellitus is a chronic metabolic disorder characterized by elevated levels of blood glucose (blood sugar) over time. This typically occurs when the pancreas does not produce enough insulin, or when the body cannot effectively use the insulin it produces. Insulin is a hormone that regulates blood glucose, allowing cells to absorb glucose from the blood and use it for energy.

### Types of Diabetes
Diabetes is classified into several main types:
* **Type 1 Diabetes:** An autoimmune condition where the immune system attacks and destroys insulin-producing beta cells in the pancreas, leading to severe insulin deficiency. This type is often diagnosed in children and young adults.
* **Type 2 Diabetes:** The most common form of diabetes, characterized by insulin resistance, where cells fail to respond properly to insulin. The pancreas may also produce insufficient insulin. It is strongly associated with lifestyle factors such as obesity and lack of physical activity.
* **Prediabetes:** A metabolic condition where blood glucose levels are higher than normal but not yet high enough to be classified as Type 2 diabetes.

### Risk Factors
Several clinical and demographic factors increase the risk of developing diabetes:
* **Age:** The risk of diabetes, especially Type 2, increases significantly with age due to metabolic changes.
* **Obesity & Weight:** Excess body weight, particularly visceral fat accumulation in the abdominal region, is a primary driver of insulin resistance.
* **Physical Inactivity:** Regular physical activity improves insulin sensitivity and glucose uptake by muscles.
* **Unhealthy Diet:** High consumption of refined carbohydrates, sugars, and saturated fats contributes to metabolic dysfunction.
* **Family History:** Genetic factors play a significant role in individual susceptibility to diabetes.
* **Smoking:** Tobacco use increases oxidative stress, inflammation, and insulin resistance.

### Health Indicators & Reference Ranges
To properly analyze and interpret health data, it is crucial to understand the normal reference ranges for key biological indicators:

#### Blood Glucose Levels

HbA1c measures the average blood glucose levels over the past 2–3 months.
| Status | HbA1c Value |
|---|---|

| **Prediabetes** | 5.7% – 6.4% |
| **Diabetes** | ≥ 6.5% |

#### Body Mass Index (BMI)

#### Blood Pressure

|---|---|

## Dataset Features & Clinical Explanations

### Demographic & Identification Features

#### `patient_id`
* **Description:** A unique identifier assigned to each patient.
* **Type:** Integer (Nominal)
* **Clinical Significance:** Used strictly to distinguish individual records. It has no predictive value and will be excluded from modeling.

#### `age`
* **Description:** Age of the patient in years (Range: 18 – 90).
* **Type:** Integer (Ratio)
* **Clinical Significance:** The risk of metabolic disorders and insulin resistance increases with age due to progressive decline in pancreatic beta-cell function and metabolic rate.

#### `gender`
* **Description:** Biological sex of the patient.
* **Type:** Categorical (Nominal)
* **Values:** `Male`, `Female`, `Other`
* **Clinical Significance:** Gender-specific differences exist in fat distribution (android vs. gynoid) and hormonal profiles (estrogen vs. testosterone), which influence insulin sensitivity.

#### `ethnicity`
* **Description:** Ethnic background of the patient.
* **Type:** Categorical (Nominal)
* **Values:** `White`, `Hispanic`, `Black`, `Asian`, `Other`
* **Clinical Significance:** Genetic predisposition and lifestyle differences among ethnic groups influence their susceptibility to diabetes.

---

### Socio-Economic Factors

#### `education_level`
* **Description:** Highest formal education completed by the patient.
* **Type:** Categorical (Ordinal)
* **Values:** `No formal education`, `Highschool`, `Graduate`, `Postgraduate`
* **Clinical Significance:** Higher education is often associated with greater health literacy, healthier lifestyle choices, and better access to medical care.

#### `income_level`
* **Description:** Household income level.
* **Type:** Categorical (Ordinal)
* **Values:** `Low`, `Lower-Middle`, `Middle`, `Upper-Middle`, `High`
* **Clinical Significance:** Income impacts the affordability of nutrient-dense food, recreational activities, and preventative healthcare.

#### `employment_status`
* **Description:** Employment status of the patient.
* **Type:** Categorical (Nominal)
* **Values:** `Employed`, `Unemployed`, `Retired`, `Student`
* **Clinical Significance:** Employment conditions influence physical activity levels, stress levels, and daily routines.

---

### Lifestyle Factors

#### `smoking_status`
* **Description:** Current smoking habits.
* **Type:** Categorical (Nominal)
* **Values:** `Never`, `Former`, `Current`
* **Clinical Significance:** Smoking triggers systemic inflammation, oxidative stress, and abdominal fat accumulation, directly increasing the risk of insulin resistance.

#### `alcohol_consumption_per_week`
* **Description:** Weekly alcohol consumption (drinks per week, Range: 0 – 30).
* **Type:** Float (Ratio)
* **Clinical Significance:** Excessive alcohol intake is linked to chronic pancreatitis, liver dysfunction, and disrupted glucose homeostasis.

#### `physical_activity_minutes_per_week`
* **Description:** Weekly duration of physical exercise in minutes (Range: 0 – 600).
* **Type:** Integer (Ratio)
* **Clinical Significance:** Physical activity enhances glucose transport (GLUT4 translocation) in skeletal muscle and improves insulin sensitivity independent of weight loss. The recommended target is at least 150 minutes per week.

#### `diet_score`
* **Description:** A score evaluating dietary habits (Range: 0 – 10).
* **Type:** Integer (Ordinal)
* **Clinical Significance:** Higher scores indicate a diet rich in fiber, micronutrients, and healthy fats, which helps regulate blood glucose and lipid levels.

#### `sleep_hours_per_day`
* **Description:** Average daily sleep duration in hours (Range: 3 – 12).
* **Type:** Float (Ratio)
* **Clinical Significance:** Sleep deprivation alters cortisol levels, increases appetite-stimulating hormones (ghrelin), and impairs peripheral insulin sensitivity. The optimal range is 7 – 9 hours per day.

#### `screen_time_hours_per_day`
* **Description:** Daily average screen time in hours (Range: 0 – 12).
* **Type:** Float (Ratio)
* **Clinical Significance:** Prolonged screen time is a marker for sedentary behavior, which leads to weight gain and poor metabolic health.

---

### Medical & Family History

#### `family_history_diabetes`
* **Description:** Family history of diabetes.
* **Type:** Binary (Nominal)
* **Values:** `0` (No), `1` (Yes)
* **Clinical Significance:** Strong hereditary factor. Individuals with immediate family history have a significantly higher risk of developing the disease.

#### `hypertension_history`
* **Description:** Personal history of hypertension (high blood pressure).
* **Type:** Binary (Nominal)
* **Values:** `0` (No), `1` (Yes)
* **Clinical Significance:** Hypertension is a key component of metabolic syndrome and often co-occurs with insulin resistance and dyslipidemia.

#### `cardiovascular_history`
* **Description:** Personal history of cardiovascular disease.
* **Type:** Binary (Nominal)
* **Values:** `0` (No), `1` (Yes)
* **Clinical Significance:** Cardiovascular diseases share common underlying pathophysiological mechanisms with diabetes, primarily endothelial dysfunction and atherosclerosis.

---

### Physical Measurements

#### `bmi`
* **Description:** Body Mass Index ($kg/m^2$, Range: 15 – 45).
* **Type:** Float (Ratio)
* **Clinical Significance:** Elevated BMI (specifically obese status $\ge 30$) is a major risk factor for type 2 diabetes due to lipid accumulation in non-adipose tissues (ectopic fat).

#### `waist_to_hip_ratio`
* **Description:** Ratio of waist circumference to hip circumference (Range: 0.7 – 1.2).
* **Type:** Float (Ratio)
* **Clinical Significance:** This ratio is a strong indicator of central (visceral) adiposity, which is more metabolically active and directly linked to insulin resistance than peripheral fat.

#### `systolic_bp`
* **Description:** Systolic blood pressure in mmHg (Range: 90 – 180).
* **Type:** Integer (Interval)
* **Clinical Significance:** High blood pressure indicates cardiovascular stress and is associated with arterial stiffness, a common feature in diabetic patients.

#### `diastolic_bp`
* **Description:** Diastolic blood pressure in mmHg (Range: 60 – 120).
* **Type:** Integer (Interval)
* **Clinical Significance:** Represents the pressure in arteries when the heart rests, another key component in assessing vascular health.

#### `heart_rate`
* **Description:** Resting heart rate in beats per minute (Range: 50 – 120).
* **Type:** Integer (Ratio)
* **Clinical Significance:** Elevated resting heart rate can indicate autonomic nervous system imbalance or poor cardiovascular fitness, both of which correlate with metabolic risks.

---

### Lipid Panel Features

#### `cholesterol_total`
* **Description:** Total serum cholesterol level in mg/dL (Range: 120 – 300).
* **Type:** Float (Ratio)
* **Clinical Significance:** An essential indicator of lipid profile. High total cholesterol increases cardiovascular risk.

#### `hdl_cholesterol`
* **Description:** High-Density Lipoprotein ("good") cholesterol level in mg/dL (Range: 20 – 100).
* **Type:** Float (Ratio)
* **Clinical Significance:** HDL aids in reverse cholesterol transport. Low levels are associated with insulin resistance and elevated risk of cardiovascular events.

#### `ldl_cholesterol`
* **Description:** Low-Density Lipoprotein ("bad") cholesterol level in mg/dL (Range: 50 – 200).
* **Type:** Float (Ratio)
* **Clinical Significance:** High LDL levels promote arterial plaque formation (atherosclerosis) and are highly prevalent in individuals with diabetic dyslipidemia.

#### `triglycerides`
* **Description:** Serum triglyceride levels in mg/dL (Range: 50 – 500).
* **Type:** Float (Ratio)
* **Clinical Significance:** Elevated triglycerides are closely linked to insulin resistance, obesity, and cardiovascular diseases.

---

### Glucose & Insulin Markers

#### `glucose_fasting`
* **Description:** Blood glucose level after fasting for at least 8 hours in mg/dL (Range: 70 – 250).
* **Type:** Float (Ratio)
* **Clinical Significance:** Primary diagnostic marker for diabetes. Persistent fasting glucose $\ge 126$ mg/dL indicates diabetes.

#### `glucose_postprandial`
* **Description:** Blood glucose level measured 2 hours after a meal in mg/dL (Range: 90 – 350).
* **Type:** Float (Ratio)
* **Clinical Significance:** Evaluates the body's post-meal glucose clearance rate. Values $\ge 200$ mg/dL signify diabetes.

#### `insulin_level`
* **Description:** Serum insulin level in $\mu U/mL$ (Range: 2 – 50).
* **Type:** Float (Ratio)
* **Clinical Significance:** Elevated fasting insulin levels are a clinical hallmark of insulin resistance, indicating the pancreas is compensating for reduced cell sensitivity.

#### `hba1c`
* **Description:** Glycated hemoglobin percentage (Range: 4 – 14%).
* **Type:** Float (Ratio)
* **Clinical Significance:** The gold standard for diabetes monitoring, reflecting long-term glycemic control. A level of $\ge 6.5\%$ is diagnostic for diabetes.

---

### Target & Risk Indicators

#### `diabetes_risk_score`
* **Description:** Calculated risk score for diabetes (Range: 0 – 100).
* **Type:** Integer (Interval)
* **Clinical Significance:** A composite risk score derived from medical and lifestyle variables. This feature leaks the target status and will be dropped.

#### `diabetes_stage`
* **Description:** Categorical classification of diabetes stage.
* **Type:** Categorical (Nominal)
* **Values:** `No Diabetes`, `Pre-Diabetes`, `Type 1`, `Type 2`, `Gestational`
* **Clinical Significance:** Reflects the disease severity classification. It is redundant and leaks the target, hence it will be dropped.

#### `diagnosed_diabetes`
* **Description:** Diagnostic status of diabetes.
* **Type:** Binary (Nominal)
* **Values:** `0` (No Diabetes), `1` (Diabetes)
* **Clinical Significance:** The binary target variable for classification models in this project.

## Outlier Detection and Treatment

### Outlier Detection and Handling Strategy

We utilize box plots to visualize the distribution and detect outliers across all numerical features.

**Handling Rationale:**
In clinical datasets, extreme measurements (such as high BMI, high glucose levels, or elevated insulin) represent valid biological variations and are crucial indicators of pathological conditions like diabetes. Removing these records could strip away the most informative signals for training our predictive models. To manage outliers without data loss, we:
1. Use robust scaling (`RobustScaler`) that scales features using statistics robust to outliers (median and IQR).
2. Employ machine learning algorithms that are naturally robust to extreme values, such as Random Forest, XGBoost, and CatBoost.

### Box Plots for Outlier Detection

Generate box plots for all continuous numerical features to inspect and flag outliers.


```python
# Generate box plots for continuous variables
num_cols = data.select_dtypes(include='number')
cols = [col for col in num_cols.columns if data[col].nunique() > 2]
n_cols = 4
n_rows = math.ceil(len(cols) / n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 4 * n_rows))
data[cols].plot(kind='box', subplots=True, layout=(n_rows, n_cols), ax=axes, sharex=False, sharey=False)
fig.suptitle("Box Plots of Numeric Features", y=0.98, fontsize=18, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### Insight
- **Observation:** The box plots show a large number of outliers (data points beyond 1.5 * IQR) in several variables, including `insulin_level`, `bmi`, `glucose_fasting`, and `triglycerides`.
- **Clinical Context:** Clinical datasets often exhibit high variance because extreme physiological states (such as acute hyperglycemia or severe obesity) are actual pathological manifestations.
- **Modeling Impact:** Since these outliers represent genuine clinical events rather than noise, they must not be removed. Using robust scaling and algorithms like CatBoost, which partition data using decision trees, ensures these extreme values do not degrade model training.
